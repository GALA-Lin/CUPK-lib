该作业要求：根据提供的数据对白葡萄酒的品质进行分析，根据Cell中的要求补全代码

```python
#读取数据
import csv
f = open("/home/aistudio/data/data351631/white_wine.csv", "r")
#补全代码
reader = csv.reader(f)  
content = []
for row in reader:
    content.append(row)
f.close()
for i in range(5):
    print(content[i])
```

```python
# 查看白葡萄酒共分为几种等级
quality_list = []
# 补全代码
for row in content[1:]:  # 跳过表头
    quality = int(row[-1])
    quality_list.append(quality)
# 用集合去重，得到所有不重复的等级
quality_count = set(quality_list)
print("白葡萄酒共有%d种等级，分别是:%r" % (len(quality_count), quality_count))
```

```python
#按酒的等级将数据集划分为7个子集，并统每种等级的数量
content_dict = {}
for row in content[1:]:
    quality = int(row[-1])
    if quality not in content_dict.keys():
        content_dict[quality] = [row]
    else:
        #补全代码
        content_dict[quality].append(row)
for key in content_dict:
    print(key,":",len(content_dict[key]))
```

```python
#统计在每个品质的样本量
number_tuple = []
for key, value in content_dict.items():
    #补全代码
    number_tuple.append( (key, len(value)) ) #元组
print (number_tuple)
```

```python
#求每个数据集中fixed acidity的均值
mean_list = []
for key, value in content_dict.items():
    sum= 0
    for row in value:
        sum += float(row[0]) #fixed acidity是第一列数据
    #补全代码
    mean_list.append( (key, sum / len(value) ) )
for item in mean_list:
    print(item[0],":",item[1])
#     print("\n")
# print(mean_list)
```

## 向列表中追加元组
### 对列表append一组信息 append(key,value)不行，而append((key,value))为什么可以？
#### list.append() 只接受 1 个参数
append() 是列表的内置方法，它的设计目的是 “向列表末尾添加一个元素”，因此调用时必须且只能传入 1 个参数
而(key, value) 是一个 元组（tuple），而元组在这里是作为 “单个容器” 存在的 —— 它把 key 和 value 这两个数据 “打包” 成了一个整体，相当于给 append() 传递了 “1 个参数”（这个参数就是包含两个数据的元组）
